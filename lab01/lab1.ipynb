```python
with open('dataset_raw.txt', 'r', encoding='utf8') as file:
    text = file.read()
```

```python
print(text[:200])
```

```python
def clean_spaces(x):
    return ' '.join(x.split())

with open('dataset_clean.txt', 'w', encoding='utf8') as file:
    file.write(clean_spaces(text))
```

```python
def split_sents(x):
    return x.replace('!', '.').replace('?', '.').split('.')

for s in split_sents(text):
    print(s)
```

```python
def get_vocabulary(x):
    return sorted(list(set(x.lower().split())))

V = get_vocabulary(text)
print(len(V))
```
